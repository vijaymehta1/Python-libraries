```python
import numpy as np
import pandas as pd
```

```python
movie = pd.read_csv('imdb-top-1000.csv')
movie.head(1)
```

```python
genre = movie.groupby('Genre')
```

```python
genre.sum().head(1)
```

 find the top 3 genres by total earning

```python
movie.groupby('Genre')['Gross'].sum().sort_values(ascending=False).head(3)
```

 find the genre with highest avg IMDB rating

```python
movie.groupby('Genre')['IMDB_Rating'].mean().sort_values(ascending=False).head(1)
```

find director with most popularity

```python
movie.groupby('Director')['No_of_Votes'].sum().sort_values(ascending=False).head(1)
```

find the highest rated movie of each genre

```python
movie.groupby('Genre')['IMDB_Rating'].max()
```

find number of movies done by each actor

```python
movie.groupby('Star1')['Series_Title'].count().sort_values(ascending=False)
```
