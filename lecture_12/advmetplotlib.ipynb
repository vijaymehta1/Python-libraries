```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns
```

Colored Scatterplots

```python
iris = pd.read_csv('iris.csv',encoding_errors='ignore')
plt.scatter(iris['SepalWidthCm'],iris['PetalWidthCm'])
iris
```

```python
iris['Species'] = iris['Species'].replace({'Iris-setosa':0,'Iris-versicolor':1,'Iris-virginica':2})
```

```python
# cmap and alpha 
plt.scatter(iris['SepalWidthCm'],iris['PetalWidthCm'],c=iris['Species'],cmap="jet", alpha=0.8)
plt.xlabel('SepalWidthCm')
plt.ylabel('PetalWidthCm')
plt.colorbar()
plt.show()
```

Plot size

```python
plt.figure(figsize=(15,7))
plt.scatter(iris['SepalWidthCm'],iris['PetalWidthCm'],c=iris['Species'],cmap="jet", alpha=0.8)
plt.xlabel('SepalWidthCm')
plt.ylabel('PetalWidthCm')
plt.colorbar()
plt.show()
```

```python
batter = pd.read_csv('batter.csv')
batter.head(1)
```

```python
batter.shape
```

```python
sample_df = batter.head(100).sample(20,random_state=5)
sample_df
```

```python
plt.figure(figsize=(15,7))
plt.scatter(sample_df['strike_rate'],sample_df['avg'],s = sample_df['runs']/10)
for i in range(sample_df.shape[0]):
    plt.text(sample_df['strike_rate'].values[i],sample_df['avg'].values[i],sample_df['batter'].values[i])
```

Horizontal and Vertical lines

```python
plt.figure(figsize=(15,7))
plt.scatter(sample_df['strike_rate'],sample_df['avg'],s = sample_df['runs']/10)
plt.axhline(35,color = 'red')
plt.axvline(140,color = 'green')
for i in range(sample_df.shape[0]):
    plt.text(sample_df['strike_rate'].values[i],sample_df['avg'].values[i],sample_df['batter'].values[i])
```

```python
fig ,ax = plt.subplots(figsize = (15,6))
ax.scatter(batter['avg'],batter['strike_rate'])
ax.set_xlabel('runs')
ax.set_ylabel('strike rate ')
ax.set_title('avg vs strike rate ')
fig.show()
```

```python
# batter dataset
fig,ax = plt.subplots(nrows=2,ncols=1,sharey=True,figsize =(20,12))
ax[0].scatter(batter['runs'],batter['strike_rate'],color='red')
ax[1].scatter(batter['avg'],batter['strike_rate'],color = 'green')
ax[0].set_title('runs Vs Strike Rate')
ax[0].set_ylabel('Strike Rate')


ax[1].set_title('Avg Vs strike rate ')
ax[1].set_ylabel('Runs')
ax[1].set_xlabel('Avg')
```

```python
fig, ax = plt.subplots(nrows=2,ncols=2,figsize=(10,10))

ax[0,0].scatter(batter['runs'],batter['strike_rate'],color='red')
ax[0,1].scatter(batter['avg'],batter['runs'])
ax[1,0].hist(batter['avg'])
ax[1,1].hist(batter['runs'])
```

```python
fig = plt.figure()

ax1 = fig.add_subplot(2,2,1)
ax1.scatter(batter['avg'],batter['strike_rate'],color='red')

ax2 = fig.add_subplot(2,2,2)
ax2.hist(batter['runs'])

ax3 = fig.add_subplot(2,2,3)
ax3.hist(batter['avg'])
```
