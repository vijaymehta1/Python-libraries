2. Advanced Indexing

```python
a=np.arange(24).reshape(6,4)
print(a)
```

```python
fancy_row=a[[0,2,4,5],:]
print(fancy_row)
```

```python
fancy_col=a[:,[1,3]]
print(fancy_col)
```

B. Boolean Indexing


```python
b=np.random.randint(1,100,24).reshape(6,4)
print(b)
```

```python
mask = b>50
print(mask)
```

```python
result=a[mask]
print(result)
```
