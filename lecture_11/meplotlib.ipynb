```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sb 
```

2D Line plot

Bivariate Analysis

categorical -> numerical and numerical -> numerical

Use case - Time series data

```python
# plotting a simple function
price = [ 22000,23000,25000,20000,30000]
year = [2015,2016,2017,2018,2019]
plt.plot(year,price)
```

```python
# from a pandas dataframe
batman = pd.read_csv('sharma-kohli.csv')
batman.head(1)
```

```python
plt.plot(batman['index'],batman['RG Sharma'],marker = 'D',label = 'virat')
plt.plot(batman['index'],batman['V Kohli'],marker = 'o', label = 'rohit')
plt.title('rohit sharma vs virat kohli career comparison')
plt.ylabel('runs scored')
plt.xlabel('season')
# legend -> location
plt.legend(loc = 'upper right')

# grid
plt.grid()
```

```python
# limiting axes
price = [48000,54000,57000,49000,47000,45000,4500000]
year = [2015,2016,2017,2018,2019,2020,2021]

plt.plot(year,price)
plt.ylim(0,60000)
```

```python
price = [48000,54000,57000,49000,47000,45000,4500000]
year = [2015,2016,2017,2018,2019,2020,2021]

plt.plot(year,price)
plt.xlim(2015,2020)
```

Scatter Plots --------->

Bivariate Analysis

numerical vs numerical

Use case - Finding correlation

```python
# plt.scatter simple function
x =  np.linspace(-10,10,50)
y =  x*10+3 +np.random.randint(0,300,50)
plt.scatter(x,y)
plt.title('x vs y scatter graph')
```

```python
# plt.scatter on pandas data
df = pd.read_csv('batter.csv')
df.head(2)
```

```python
plt.scatter(df['avg'],df['strike_rate'])
```

```python
tips = sb.load_dataset('tips')
# slower
plt.scatter(tips['total_bill'],tips['tip'],s= tips['size']*10)
```

```python
# scatterplot using plt.plot
# faster
plt.plot(tips['total_bill'],tips['tip'],'o')
```

```python
# simple bar chart
students = [23,44,56,57,101]
color = ['blue','pink','yellow','red','orange']
plt.bar(color,students,color = 'blue')
```
